##Python Data Types


---

Commonly used data types in Python include:


*   String: A sequence of characters - `"hello"`
*   Int: A whole number - `10`
*   Float: A decimal number - `2.5`
*   List: An ordered list of items - `[1, 2, 3, 4]`
*   Dict: A set of key-value pairs - `{"key1":"value1", "key2":"value2", "key3":"value3"}`
*   Bool: A true/false value - `True`


---


Variables can be assigned in Python using the `=` operator

`string_var = "hello"`

`int_var = 10`


###Strings

Strings can be defined using either single quotes ('...') or double quotes ("...")

````python
#Create a string with single quotes
str1 = 'Hello'

#Create a string with double quotes
str2 = "Goodbye"

#Print the strings
print(str1, str2)
````

###Ints

Ints can be defined using a number or by casting another data type using `int()`

````python
#Create an int
int1 = 10

#Cast a string to an int
int2 = int("10")

#Cast a float to an int (will truncate the number)
int3 = int(10.5)

#Print the ints
print(int1, int2, int3)
````

###Floats

Floats can be defined using a number or by casting another data type using `float()`

````python
#Create an float
float1 = 10.5

#Cast a string to an float
float2 = float("10.5")

#Cast a float to an float
float3 = float(10)

#Print the floats
print(float1, float2, float3)
````

###Lists

Lists are defined using square brackets ([...]) and can contain any datatype

````python
#Create a list of ints
list1 = [1, 2, 3, 4, 5]

#Create a list of strings
list2 = ["Apple", "Ball", "Cat"]

#Create a list of lists
list3 = [list1, list2]

#Print the lists
print(list1, list2, list3)
````

Elements of a list can be accessed with the square brackets

````python
#Access the first element of list1
print(list1[0])

#Access the second element of list2
print(list2[1])
````

The length of a list can be found with `len()`

````python
#Get the length of list1
print(len(list1))
````

###Dicts

Dicts are defined using curly braces ({...}) and key-value pairs (key:value)

````python
#Create a dict with string keys and values
dict1 = {"key1":"value1", "key2":"value2", "key3":"value3", "key4":"value4", "key5":"value5"}

#Create a dict with string keys and int values
dict2 = {"key1":1, "key2":2, "key3":3, "key4":4, "key5":5}

#Print the dicts
print(dict1, dict2)
````

Values in a dict can be accessed using its key

````python
#Accessing the value at key1 in dict1
print(dict1["key1"])

#Accessing the value at key4 in dict2
print(dict2["key4"])
````

The keys and values of a dict can be found using `.keys()` and `.values()`

````python
#Get the keys of dict1
print(dict1.keys())

#Get the values of dict2
print(dict2.values())
````

###Bools

Bools can be either `True` or `False`

````python
#Create a true bool
bool1 = True

#Create a false bool
bool2 = False

#Print the bools
print(bool1, bool2)
````

##Python Conditional Statements

Conditional statements are used to branch code based on a condition



---

General structure of a conditional statement:


```
if condition is met:
  # Do this
```


---

Commonly used operators in conditional statements:
*   `a == b` - True if a is equal to b
*   `a != b` - True if a is not equal to b
*   `a > b` - True if a is greater than b
*   `a >= b` - True if a is greater than or equal to b
*   `a < b` - True if a is less than b
*   `a <= b` - True if a is less than or equal to b
*   `a and b` - True if a and b are true
*   `a and b` - True if a or b are true
*   `not a` - True if a is not true

The `if` keyword is used to create a conditional statement

````python
#Create a simple if-statement
if 10 > 0:
  print("10 is greater than 0")
````

The `elif` keyword can be used to add multiple conditions to a statement

````python
#Create an if-statement with multiple conditions
if 10 > 15:
  print("10 is greater than 15")
elif 10 > 5:
  print("10 is greater than 5")
````

The `else` keyword is used as a final condition if none other are met

````python
#Create an if-statement with multiple conditions
if 10 > 20:
  print("10 is greater than 20")
elif 10 > 15:
  print("10 is greater than 15")
else:
  print("10 is not greater than 20 or 15")
````

Other operators in if-statements

````python
#Create ints
x = 10
y = 20

#Create an if-statement with the == operator
if x == 10:
  print("x equals 10")

#Create an if-statement with the >= operator
if x >= 5:
  print("x is greater than or equal to 10")

#Create an if-statement with the < operator
if x < 15:
  print("x is less than 15")

#Create an if-statement with the == operator
if x != 15:
  print("x is not equal to 15")

#Create an if-statement with the and operator
if x == 10 and y == 20:
  print("x equals 10 and y equals 20")

#Create an if-statement with the or operator
if x == 15 or y == 20:
  print("x equals 10 or y equals 20")
````

Bools can be used in conditional statements

````python
#Create a true bool
bool1 = True

#Create an if-statement with bool1 as the condition
if bool1:
  print("bool1 is true")
else:
  print("bool1 is false")
````

##Python Loops

Loops are used to make piece of code repeat multiple times


---

The two types of loops are for loops and while loops

###For Loops

For loops can used for repeating a set number of times by using the `range()` function

````python
#Create a for loop the repeats 10 times
for i in range(10):
  print(i)

print()

#Create a for loop the repeats 10 times, starting at 10
for i in range(10, 20):
  print(i)
````

For loops can also be used to iterate over a list

````python
#Create a list of strings
list1 = ["Apple", "Ball", "Cat"]

#Create a for loop that iterates over list1
for string in list1:
  print(string)
````

###While Loops

While loops are used to repeat until a condition is met

````python
#Create an int
x = 0

#Create a while loop that reapeats as long as bool1 is true
while x < 10:
  print(x)
  x += 1
````

##Python Functions

Functions are used to do a task multiple times without needing to rewrite the code each time


---


Basic function structure:


```
def function_name():
  # Code to do something
```

Basic function call:

`function_name()`


The contents of a function must be indented from the fucntion declaration

````python
#Create a function that prints "This is a function"
def fun():
  print("This is a function")

fun()
````

Variables can be passed to a function as parameters

````python
#Create a function that prints the parameter that is passed to it
def print_param(param):
  print(param)

#Call the function, passing in "Hello" as the parameter
print_param("Hello")
````

Functions can also return values

````python
#Create a function that adds the two parameters and returns the result
def add(param1, param2):
  result = param1 + param2
  return result

#Print the result of calling add on 5 and 10
print(add(5, 10))
````

###Sklearn
The `sklearn` library provides tools for data analysis and machine learning

````python
#Import the linear regression function from sklearn
from sklearn.linear_model import LinearRegression
import numpy as np
````

Formatting data

````python
#Get X and y variables from dataframe
X = merged_receiving["Receptions"].values.reshape(-1, 1)
y = merged_receiving["Yards"].values
````

Training a model

````python
#Create a linear regression model using receptions and yards
model = LinearRegression()
model.fit(X, y)
````

Making predictions

````python
#Print the model coefficient
print(model.coef_[0])

#Print the model intercept
print(model.intercept_)
````

````python
#Make a prediction using the model
prediction = model.predict(np.array([100]).reshape(-1, 1))[0]

#Print the result of the prediction
print("If this receiver were to get 100 receptions, he would be expected to have", prediction, "yards")
````
